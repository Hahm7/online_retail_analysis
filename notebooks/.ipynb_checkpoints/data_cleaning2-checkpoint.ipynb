## How this notebook works

Cleaning is one **staged CTE chain**. Each stage fixes exactly one issue and feeds the next:

```
raw_retail
  └─ deduped                                  remove exact duplicate rows
      └─ cancellation_tag                     add is_cancellation flag
          └─ base_transactions                drop Price <= 0
              └─ stock_code_..._length_adjust drop short codes and admin codes
                  └─ stock_code_without_gift  drop gift voucher codes
                      └─ stock_code_variant_fix  uppercase stock codes
                          └─ description_trimmed  trim descriptions
                              └─ description_upper  uppercase descriptions
                                  └─ guest_tag       add is_guest flag
                                      └─ revenue_derive  Revenue = Quantity × Price
                                          └─ cleaned        add date features
```

What to know before reading the cells:

1. **Every cell repeats the full chain up to its own stage.** This is on purpose. Each cell runs on its own without needing to run the cells above it first. Section 12 puts the finished chain together once as a view.
2. **A "remaining issues" cell follows each stage and counts every problem still left.** Many rows carry more than one issue at a time. These counts show what the stage fixed.

Each decision to remove data is explained in writing before the query that does it. Section 11 shows how many rows every stage removed.

## 1. Setup

Load both excel sheets, concatenate them into a single dataframe and register that dataframe with duckDB as the table `raw_retail`

````python
import duckdb
import pandas as pd

df1 = pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2009-2010')
df2 = pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2010-2011')
raw = pd.concat([df1, df2], ignore_index=True)


print(f"Loaded {len(raw):,} rows")
````

## 2. Standardise column names

`Customer ID` is renamed to `CustomerID` so it no longer needs quotes in every query

````python
raw = raw.rename(columns={'Customer ID': 'CustomerID'})

con = duckdb.connect()
con.register('raw_retail', raw)

print("Columns:", list(raw.columns))
print(f"Rows: {len(raw):,}")
````

## 3. Remove exact duplicates

**Issue found in audit:** 34,335 rows identical across every column, including the timestamp. They are 2% of rows and £431,717 of revenue.

**Decision:** remove them. With no line-item ID there is no way to tell a genuine repeated line from an extraction error and the volume is small enough that the risk of double counting outweighs the risk of dropping real sales.

**Stage:** `deduped`

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    )
    SELECT COUNT(*) AS no_duplication_total FROM deduped;
""").fetchdf()
````

## 4. Flag cancellations

**Issue found in audit:** 19,494 rows have a `C`-prefixed invoice.

**Decision:** flag them. Cancellations are needed to measure the return rate in the analysis notebook, so they get an `is_cancellation` column and every downstream analysis filters on it explicitly.

**Stage:** `cancellation_tag`

````python
# Remaining data quality issues
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * FROM raw_retail
    )
    SELECT
        COUNT(*) total_rows,
        SUM(CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' OR Quantity < 0 THEN 1 ELSE 0 END) AS cancellations,
        SUM(CASE WHEN Price <= 0 THEN 1 ELSE 0 END) AS invalid_price,
        SUM(CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END) AS null_customer_id,
        SUM(CASE WHEN Description IS NULL THEN 1 ELSE 0 END) AS null_description
    FROM deduped;
""").fetchdf()
````

````python
# Flag Cancellations
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    )
    SELECT COUNT(*) AS rows_after_tag FROM cancellation_tag;
""").fetchdf()
````

````python
# Remaining data quality issues after flagging cancellations
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    )
    SELECT
        COUNT(*) total_rows,
        SUM(CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' OR Quantity < 0 THEN 1 ELSE 0 END) AS flagged_cancellations,
        SUM(CASE WHEN Price <= 0 THEN 1 ELSE 0 END) AS invalid_price,
        SUM(CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END) AS null_customer_id,
        SUM(CASE WHEN Description IS NULL THEN 1 ELSE 0 END) AS null_description
    FROM cancellation_tag;
""").fetchdf()
````

## 5. Remove zero and negative prices

**Issue found in audit:** 6,207 rows with zero or negative price. Negative prices are bad-debt adjustments while zero prices have no documented meaning.

**Decision:** remove. Neither represents a sale and both would distort revenue and average order value.

**Stage:** `base_transactions`

````python
# remove zero/negative prices
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    )
    SELECT COUNT(*) AS rows_after_filter FROM base_transactions;
""").fetchdf()
````

````python
# Remaining data quality issues after removing zero and negative prices
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    )
    SELECT
        COUNT(*) total_rows,
        SUM(CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' OR Quantity < 0 THEN 1 ELSE 0 END) AS flagged_cancellations,
        SUM(CASE WHEN Price <= 0 THEN 1 ELSE 0 END) AS invalid_price,
        SUM(CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END) AS null_customer_id,
        SUM(CASE WHEN Description IS NULL THEN 1 ELSE 0 END) AS null_description
    FROM base_transactions;
""").fetchdf()
````

> **Note:** removing zero and negative prices also cleared every null `Description`. The two issues sat on the same rows.

## 6. Clean stock codes

Stock codes are usually 5–6 characters of digits or digits plus letters.

### 6.1 Which codes are not products?

````python
# Stock Code Distribution
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    )
    SELECT
        LENGTH(CAST(StockCode AS VARCHAR)) AS code_length,
        COUNT(*) AS rows,
        COUNT(DISTINCT StockCode) AS distinct_codes,
        ANY_VALUE(StockCode) AS example
    FROM 
        base_transactions
    GROUP BY 
        code_length
    ORDER BY 
        rows DESC
""").fetchdf()
````

````python
# Stock Codes without numbers
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    )
    SELECT
        CAST(StockCode AS VARCHAR) AS code,
        ANY_VALUE(Description) AS description,
        COUNT(*) AS rows,
        ROUND(MIN(Price), 2) AS min_price,
        ROUND(MAX(Price), 2) AS max_price
    FROM 
        base_transactions
    WHERE 
        NOT REGEXP_MATCHES(CAST(StockCode AS VARCHAR), '[0-9]')
    GROUP BY 
        code
    """).fetchdf()
````

**Decision:** codes with no digits are not products, they are ledger entries such as `BANK CHARGES`, `ADJUST` and `AMAZONFEE`. Drop those and keep only codes of 5 characters or more.

**Stage:** `stock_code_without_numbers_length_adjust`

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    )
    SELECT
        LENGTH(CAST(StockCode AS VARCHAR)) AS code_length,
        COUNT(*) AS rows,
        COUNT(DISTINCT StockCode) AS distinct_codes,
        ANY_VALUE(StockCode) AS example
    FROM 
        stock_code_without_numbers_length_adjust
    GROUP BY 
        code_length
    ORDER BY 
        rows DESC
    """).fetchdf()
````

Next, codes containing `gift`.

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    )
    SELECT
        *
    FROM 
        stock_code_without_numbers_length_adjust
    WHERE
        StockCode LIKE '%gift%'
    """).fetchdf()
````

**Decision:** `gift` codes are voucher redemptions. Drop them.

**Stage:** `stock_code_without_gift`

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    ),
    stock_code_without_gift AS (
        SELECT
            *
        FROM 
            stock_code_without_numbers_length_adjust
        WHERE
            StockCode NOT LIKE '%gift%'
    )
    SELECT
        *
    FROM
        stock_code_without_gift
    WHERE
        Description LIKE '%dot%'
    """).fetchdf()
````

### 6.2 Case variants

**Issue found in audit:** the same product can appear under both `85123a` and `85123A`, splitting its sales across stock codes.

How many codes still contain a lowercase letter after the removals above?

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    ),
    stock_code_without_gift AS (
        SELECT
            *
        FROM 
            stock_code_without_numbers_length_adjust
        WHERE
            StockCode NOT LIKE '%gift%'
    )
    SELECT 
        COUNT(DISTINCT StockCode) AS raw_distinct
    FROM
        stock_code_without_gift
    WHERE 
        StockCode != UPPER(StockCode)
""").fetchdf()
````

And how many of those have an uppercase variant?

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    ),
    stock_code_without_gift AS (
        SELECT
            *
        FROM 
            stock_code_without_numbers_length_adjust
        WHERE
            StockCode NOT LIKE '%gift%'
    )
    SELECT
        UPPER(CAST(StockCode AS VARCHAR)) AS stock_code_upper,
        COUNT(DISTINCT StockCode) AS variants,
        COUNT(*) AS total_rows
    FROM 
        stock_code_without_gift
    GROUP BY 
        stock_code_upper
    HAVING 
        COUNT(DISTINCT StockCode) > 1
    ORDER BY 
        total_rows DESC
""").fetchdf()
````

**Decision:** 170 of the 173 remaining lowercase codes have an uppercase variant, so uppercase every stock code. The query below applies the fix and then re-runs the variant check where an empty result means no code is duplicated by case any more.

**Stage:** `stock_code_variant_fix`

````python
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    ),
    stock_code_without_gift AS (
        SELECT
            *
        FROM 
            stock_code_without_numbers_length_adjust
        WHERE
            StockCode NOT LIKE '%gift%'
    ),
    stock_code_variant_fix AS (
        SELECT
            Invoice,
            UPPER(CAST(StockCode AS VARCHAR)) AS StockCode,
            Description,
            Quantity,
            InvoiceDate,
            Price,
            CustomerID,
            Country
        FROM
            stock_code_without_gift
    )
    SELECT
        UPPER(CAST(StockCode AS VARCHAR)) AS stock_code_upper,
        COUNT(DISTINCT StockCode) AS variants,
        COUNT(*) AS total_rows
    FROM 
        stock_code_variant_fix
    GROUP BY 
        stock_code_upper
    HAVING 
        COUNT(DISTINCT StockCode) > 1
    ORDER BY 
        total_rows DESC
""").fetchdf()
````

Split the affected codes into formatting noise versus genuine relabelling.

````python
# Out of the total how many are formatting noise (whitespace) vs relabelling
con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    ),
    stock_code_without_gift AS (
        SELECT
            *
        FROM 
            stock_code_without_numbers_length_adjust
        WHERE
            StockCode NOT LIKE '%gift%'
    ),
    stock_code_variant_fix AS (
        SELECT
            Invoice,
            UPPER(CAST(StockCode AS VARCHAR)) AS StockCode,
            Description,
            Quantity,
            InvoiceDate,
            Price,
            CustomerID,
            Country
        FROM
            stock_code_without_gift
    ),
    normalised AS (
        SELECT
            StockCode,
            Description,
            UPPER(TRIM(Description)) AS description_clean
        FROM stock_code_variant_fix
    ),
    summary AS (
        SELECT
            StockCode,
            COUNT(DISTINCT Description) AS raw_variants,
            COUNT(DISTINCT description_clean) AS clean_variants
        FROM normalised
        GROUP BY StockCode
        HAVING COUNT(DISTINCT Description) > 1
    )
    SELECT
        SUM(CASE WHEN clean_variants = 1 THEN 1 ELSE 0 END) AS formatting_only,
        SUM(CASE WHEN clean_variants > 1 THEN 1 ELSE 0 END) AS genuine_relabel,
        COUNT(*) AS total_affected
    FROM summary
""").fetchdf()
````

> **Insight — `Description` is not a reliable grouping key**
>
> * 624 stock codes still carry multiple descriptions at this point.
> * Only 25 of those are formatting differences. The remaining 599 are genuine relabelling — e.g. *BLUE POLKADOT GARDEN PARASOL* vs *BLUE WHITE SPOTS GARDEN PARASOL* for the same code.
> * `TRIM` and `UPPER` cannot resolve a relabelling. **Decision: normalise the formatting anyway but group products by `StockCode` and `Description` in all analysis.**

### 7.2 Trim whitespace

Saved as a pandas DataFrame to avoid having to write a long CTE chain everytime.

````python
description_trimmed = con.execute("""
    WITH deduped AS (
        SELECT DISTINCT * 
        FROM raw_retail
    ),
    cancellation_tag AS (
        SELECT *,
            CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                 OR Quantity < 0 
                 THEN 1 ELSE 0 END AS is_cancellation
        FROM deduped
    ),
    base_transactions AS (
        SELECT * 
        FROM cancellation_tag
        WHERE Price > 0
    ),
    stock_code_without_numbers_length_adjust AS (
    SELECT *
       FROM 
           (SELECT *
               FROM base_transactions
               WHERE
                   LENGTH(StockCode) >=5
            )
    WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
    ),
    stock_code_without_gift AS (
        SELECT
            *
        FROM 
            stock_code_without_numbers_length_adjust
        WHERE
            StockCode NOT LIKE '%gift%'
    ),
    stock_code_variant_fix AS (
        SELECT
            Invoice,
            UPPER(CAST(StockCode AS VARCHAR)) AS StockCode,
            Description,
            Quantity,
            InvoiceDate,
            Price,
            CustomerID,
            Country
        FROM
            stock_code_without_gift
    )
    SELECT
        Invoice,
        StockCode,
        TRIM(Description) AS Description,
        Quantity,
        InvoiceDate,
        Price,
        CustomerID,
        Country
    FROM
        stock_code_variant_fix
""").fetchdf()
````

Re-count formatting versus relabelling on the trimmed data.

````python
con.execute("""
    WITH normalised AS (
        SELECT
            StockCode,
            Description,
            UPPER(TRIM(Description)) AS description_clean
        FROM description_trimmed
    ),
    summary AS (
        SELECT
            StockCode,
            COUNT(DISTINCT Description) AS raw_variants,
            COUNT(DISTINCT description_clean) AS clean_variants
        FROM normalised
        GROUP BY StockCode
        HAVING COUNT(DISTINCT Description) > 1
    )
    SELECT
        SUM(CASE WHEN clean_variants = 1 THEN 1 ELSE 0 END) AS formatting_only,
        SUM(CASE WHEN clean_variants > 1 THEN 1 ELSE 0 END) AS genuine_relabel,
        COUNT(*) AS total_affected
    FROM summary
""").fetchdf()
````

One code still shows a formatting-only difference.

````python
con.execute("""
    WITH normalised AS (
        SELECT
            StockCode,
            Description,
            UPPER(TRIM(Description)) AS description_clean
        FROM description_trimmed
    )
    SELECT
        StockCode,
        COUNT(DISTINCT Description) AS raw_variants,
        COUNT(DISTINCT description_clean) AS clean_variants
    FROM normalised
    GROUP BY StockCode
    HAVING COUNT(DISTINCT Description) > 1 AND clean_variants = 1
    ORDER BY StockCode
""").fetchdf()
````

## 8. Tag guest transactions

**Issue found in audit:** 228,488 rows (22.7%) have no `CustomerID`.

**Decision:** flag, don't delete. Their revenue is valid and belongs in revenue totals. They are excluded only from customer level analysis via `registered_only.csv`.

**Stage:** `guest_tag`

````python
# Tag rows with no CustomerID as guest transactions
con.execute("""
    WITH guest_tag AS (
        SELECT *,
        CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END AS is_guest
        FROM description_upper
    )
    SELECT
        SUM(is_guest) AS guest_rows,
        SUM(1 - is_guest) AS registered_rows,
        COUNT(*) AS total_rows
    FROM guest_tag;
    """).fetchdf()
````

Re-count. Null descriptions were already cleared in section 5.

````python
# Remaining data quality issues after tagging guests
con.execute("""
    WITH guest_tag AS (
        SELECT *,
        CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END AS is_guest
        FROM description_upper
    )
    SELECT
        COUNT(*) total_rows,
        SUM(CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' OR Quantity < 0 THEN 1 ELSE 0 END) AS flagged_cancellations,
        SUM(CASE WHEN Price <= 0 THEN 1 ELSE 0 END) AS invalid_price,
        SUM(CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END) AS null_customer_id,
        SUM(is_guest) AS guest_total,
        SUM(CASE WHEN Description IS NULL THEN 1 ELSE 0 END) AS null_description
    FROM guest_tag;
""").fetchdf()
````

## 9. Derive revenue

`Revenue = Quantity × Price`

**Stage:** `revenue_derive`

````python
con.execute("""
    WITH guest_tag AS (
        SELECT *,
        CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END AS is_guest
        FROM description_upper
    ),
    revenue_derive AS (
        SELECT 
            *,
            Quantity * Price AS Revenue
        FROM guest_tag
    )
    SELECT
        MIN(Revenue) AS min_revenue,
        MAX(Revenue) AS max_revenue,
        ROUND(AVG(Revenue), 2) AS avg_revenue,
        ROUND(SUM(Revenue), 2) AS total_revenue,
        MIN(Quantity) AS min_quantity,
        MAX(Quantity) AS max_quantity,
        ROUND(AVG(Quantity), 2) AS avg_quantity,
        ROUND(SUM(Quantity), 2) AS total_quantity,
        MIN(Price) AS min_price,
        MAX(Price) AS max_price,
        ROUND(AVG(Price), 2) AS avg_price,
        ROUND(SUM(Price), 2) AS total_price,
        COUNT(*) total_rows
    FROM revenue_derive;
    """).fetchdf()
````

## 10. Add date features

Year, month, day of week, hour, weekend flag, truncated week and month starts.

**Stage:** `cleaned`

````python
con.execute("""
    WITH guest_tag AS (
        SELECT *,
        CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END AS is_guest
        FROM description_upper
    ),
    revenue_derive AS (
        SELECT 
            *,
            Quantity * Price AS Revenue
        FROM guest_tag
    ),
    cleaned AS (
        SELECT *,
            YEAR(InvoiceDate) AS Year,
            MONTH(InvoiceDate) AS Month,
            DAYOFWEEK(InvoiceDate) AS DayOfWeek,
            HOUR(InvoiceDate) AS Hour,
            CASE WHEN DAYOFWEEK(InvoiceDate) IN (0, 6) THEN 1 ELSE 0 END AS IsWeekend,
            DATE_TRUNC('week',  InvoiceDate) AS WeekStart,
            DATE_TRUNC('month', InvoiceDate) AS MonthStart
        FROM revenue_derive
    )
    SELECT * FROM cleaned LIMIT 5;
    """).fetchdf()
````

````python
con.execute("""
    CREATE OR REPLACE VIEW cleaned AS
    WITH deduped AS (
            SELECT DISTINCT * 
            FROM raw_retail
        ),
        cancellation_tag AS (
            SELECT *,
                CASE WHEN CAST(Invoice AS VARCHAR) LIKE 'C%' 
                     OR Quantity < 0 
                     THEN 1 ELSE 0 END AS is_cancellation
            FROM deduped
        ),
        base_transactions AS (
            SELECT * 
            FROM cancellation_tag
            WHERE Price > 0
        ),
        stock_code_without_numbers_length_adjust AS (
            SELECT *
            FROM 
                (SELECT *
                    FROM base_transactions
                    WHERE LENGTH(StockCode) >= 5
                )
            WHERE StockCode NOT IN ('BANK CHARGES', 'ADJUST', 'AMAZONFEE')
        ),
        stock_code_without_gift AS (
            SELECT *
            FROM stock_code_without_numbers_length_adjust
            WHERE StockCode NOT LIKE '%gift%'
        ),
        stock_code_variant_fix AS (
            SELECT
                Invoice,
                UPPER(CAST(StockCode AS VARCHAR)) AS StockCode,
                Description,
                Quantity,
                InvoiceDate,
                Price,
                CustomerID,
                Country,
                is_cancellation
            FROM stock_code_without_gift
        ),
        description_trimmed AS (
            SELECT
                Invoice,
                StockCode,
                TRIM(Description) AS Description,
                Quantity,
                InvoiceDate,
                Price,
                CustomerID,
                Country,
                is_cancellation
            FROM stock_code_variant_fix
        ),
        description_upper AS (
            SELECT 
                Invoice,
                StockCode,
                UPPER(Description) AS Description,
                Quantity,
                InvoiceDate,
                Price,
                CustomerID,
                Country,
                is_cancellation
            FROM description_trimmed
        ),
        guest_tag AS (
            SELECT *,
                CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END AS is_guest
            FROM description_upper
        ),
        revenue_derive AS (
            SELECT 
                *,
                Quantity * Price AS Revenue
            FROM guest_tag
        )
    SELECT *,
        YEAR(InvoiceDate) AS Year,
        MONTH(InvoiceDate) AS Month,
        DAYOFWEEK(InvoiceDate) AS DayOfWeek,
        HOUR(InvoiceDate) AS Hour,
        CASE WHEN DAYOFWEEK(InvoiceDate) IN (0, 6) THEN 1 ELSE 0 END AS IsWeekend,
        DATE_TRUNC('week',  InvoiceDate) AS WeekStart,
        DATE_TRUNC('month', InvoiceDate) AS MonthStart
    FROM revenue_derive;
""")
````

## Conventions used throughout

* **Gross vs net revenue.** "Gross" excludes cancellations and negative quantities (`is_cancellation = 0 AND Quantity > 0`). "Net" subtracts returned value. Monthly trends use gross, because a return rarely falls in the same month as the sale it reverses.
* **Trading years.** Year 1 = Dec 2009–Nov 2010, Year 2 = Dec 2010–Nov 2011. This keeps the peak trading month in the same position in each year.
* **December 2011 is excluded from time series.** This month only has 8 trading days.
* **Customer level work uses `registered`**, which excludes guest checkouts. Revenue, product and geography work uses `cleaned` which includes guests.

## 2. Revenue

### 2.1 Headline totals

**Question:** What is gross revenue, how much is lost to returns, and how many orders and customers are there?

````python
# Gross revenue, returns, net revenue, order and customer base
totals = con.execute("""
    SELECT
        ROUND(SUM(CASE WHEN is_cancellation = 0 AND Quantity > 0
                       THEN Revenue ELSE 0 END), 0) AS gross_revenue,
        ROUND(SUM(CASE WHEN is_cancellation = 1
                       THEN ABS(Revenue) ELSE 0 END), 0) AS returns,
        ROUND(SUM(CASE WHEN is_cancellation = 0 AND Quantity > 0
                       THEN Revenue ELSE 0 END)
            - SUM(CASE WHEN is_cancellation = 1
                       THEN ABS(Revenue) ELSE 0 END), 0) AS net_revenue,
        ROUND(SUM(CASE WHEN is_cancellation = 1 THEN ABS(Revenue) ELSE 0 END)
            / SUM(CASE WHEN is_cancellation = 0 AND Quantity > 0
                       THEN Revenue ELSE 0 END) * 100, 1) AS return_rate_pct,
        -- gross-only rows counted via CASE, cancellation invoices excluded
        COUNT(DISTINCT CASE WHEN is_cancellation = 0 AND Quantity > 0
                            THEN Invoice END) AS orders,
        COUNT(DISTINCT CASE WHEN is_cancellation = 0 AND Quantity > 0
                            THEN CustomerID END) AS customers
    FROM 
        cleaned
""").fetchdf()

totals
````

### 2.2 Is the business growing?

**Question:** Comparing the two complete trading years. Is revenue rising and is it driven by more orders or bigger orders?

````python
con.execute("""
    WITH sales AS (
        SELECT
            CAST(InvoiceDate AS DATE) AS invoice_date,
            Invoice,
            CustomerID,
            Revenue
        FROM 
            cleaned
        WHERE is_cancellation = 0 AND Quantity > 0
    ),
    tagged AS (
        SELECT
            *,
            CASE
                WHEN invoice_date >= DATE '2009-12-01'
                 AND invoice_date <  DATE '2010-12-01' THEN 'Year 1: Dec09-Nov10'
                WHEN invoice_date >= DATE '2010-12-01'
                 AND invoice_date <  DATE '2011-12-01' THEN 'Year 2: Dec10-Nov11'
            END AS trading_year
        FROM sales
    ),
    yearly AS (
        SELECT
            trading_year,
            ROUND(SUM(Revenue), 0) AS revenue,
            COUNT(DISTINCT Invoice) AS orders,
            COUNT(DISTINCT CustomerID) AS customers,
            ROUND(SUM(Revenue) / COUNT(DISTINCT Invoice), 2) AS avg_order_value
        FROM 
            tagged
        WHERE 
            trading_year IS NOT NULL
        GROUP BY 
            trading_year
        ORDER BY 
            trading_year
    )
    SELECT
        trading_year,
        revenue,
        orders,
        customers,
        avg_order_value,
        ROUND((revenue / LAG(revenue) OVER (ORDER BY trading_year) - 1) * 100, 1)
            AS revenue_pct_change,
        ROUND((orders / LAG(orders) OVER (ORDER BY trading_year) - 1) * 100, 1)
            AS orders_pct_change,
        ROUND((avg_order_value / LAG(avg_order_value) OVER (ORDER BY trading_year) - 1) * 100, 1)
            AS average_order_value_pct_change
    FROM 
        yearly
    ORDER BY 
        trading_year
""").fetchdf()
````

> **Insight — growth**
>
> * Revenue is broadly flat across the two years, up 2.5%.
> * Orders fell 4% while average order value rose 6.8%. Fewer transactions but each worth more, the customer base is consolidating rather than expanding.

### 2.3 Returns and cancellations

**Question:** How much revenue is reversed by returns and is the cancellation rate improving?

````python
# How much revenue is lost to returns? (Not including december, december 2010 has complete entries but december 2011 has few entries)
con.execute("""
    WITH flagged AS (
        SELECT
            CAST(InvoiceDate AS DATE) AS invoice_date,
            CASE WHEN is_cancellation = 0 AND Quantity > 0
                 THEN Revenue ELSE 0 END AS gross_revenue,
            CASE WHEN is_cancellation = 1
                 THEN ABS(Revenue) ELSE 0 END AS returned_revenue
        FROM cleaned
    ),
    tagged AS (
        SELECT
            *,
            CASE
                WHEN invoice_date >= DATE '2009-12-01'
                 AND invoice_date <  DATE '2010-12-01' THEN 'Year 1: Dec09-Nov10'
                WHEN invoice_date >= DATE '2010-12-01'
                 AND invoice_date <  DATE '2011-12-01' THEN 'Year 2: Dec10-Nov11'
            END AS trading_year
        FROM flagged
    )
    SELECT
        trading_year,
        ROUND(SUM(gross_revenue), 0) AS gross_revenue,
        ROUND(SUM(returned_revenue), 0) AS returns,
        ROUND(SUM(gross_revenue) - SUM(returned_revenue), 0) AS net_revenue,
        ROUND(SUM(returned_revenue) / SUM(gross_revenue) * 100, 1) AS return_rate_pct
    FROM tagged
    WHERE trading_year IS NOT NULL
    GROUP BY ROLLUP 
        (trading_year)
    ORDER BY 
        trading_year
""").fetchdf()
````

````python
# Cancellation Rate over time (Not including december, december 2010 has complete entries but december 2011 has few entries)
con.execute("""
    WITH monthly_orders AS (
        SELECT
            DATE_TRUNC('month', CAST(InvoiceDate AS DATE)) AS month,
            Invoice,
            is_cancellation,
            Quantity
        FROM cleaned
        WHERE CAST(InvoiceDate AS DATE) < DATE '2011-12-01'
    )
    SELECT
        month,
        COUNT(DISTINCT CASE WHEN is_cancellation = 0 AND Quantity > 0
                            THEN Invoice END) AS valid_orders,
        COUNT(DISTINCT CASE WHEN is_cancellation = 1
                            THEN Invoice END) AS cancelled_orders,
        ROUND(
            COUNT(DISTINCT CASE WHEN is_cancellation = 1 THEN Invoice END) * 100.0
            / COUNT(DISTINCT CASE WHEN is_cancellation = 0 AND Quantity > 0 THEN Invoice END)
        , 1) AS cancellation_rate_pct
    FROM monthly_orders
    GROUP BY month
    ORDER BY month
""").fetchdf()
````

````python
# Monthly total revenue, orders and customers
con.execute("""
    WITH valid_sales AS (
        SELECT *
        FROM cleaned
        WHERE is_cancellation = 0
          AND Quantity > 0
          AND CAST(InvoiceDate AS DATE) < DATE '2011-12-01'
    )
    SELECT
        DATE_TRUNC('month', CAST(InvoiceDate AS DATE)) AS month_year,
        ROUND(SUM(Revenue), 0) AS gross_revenue,
        COUNT(DISTINCT Invoice) AS orders,
        COUNT(DISTINCT CustomerID) AS customers
    FROM valid_sales
    GROUP BY 
        month_year
    ORDER BY 
        gross_revenue DESC
""").fetchdf()
````

Averaging each calendar month across both years to remove single year noise.

````python
con.execute("""
    WITH valid_sales AS (
        SELECT *
        FROM cleaned
        WHERE is_cancellation = 0
          AND Quantity > 0
          AND CAST(InvoiceDate AS DATE) < DATE '2011-12-01'
    ),
    monthly_totals AS (
        SELECT
            DATE_TRUNC('month', CAST(InvoiceDate AS DATE)) AS month_year,
            ROUND(SUM(Revenue), 0) AS gross_revenue,
            COUNT(DISTINCT Invoice) AS orders,
            COUNT(DISTINCT CustomerID) AS customers
        FROM valid_sales
        GROUP BY 
            month_year
    )
    SELECT
        strftime(month_year, '%B') AS month,
        ROUND(AVG(gross_revenue), 1) AS average_gross_revenue,
        ROUND(AVG(orders), 1) AS average_orders,
        ROUND(AVG(customers), 1) AS average_customers
    FROM
        monthly_totals
    GROUP BY
        month
    ORDER BY
        average_gross_revenue DESC 
""").fetchdf()
````

## 3. Customer segmentation (RFM)

RFM scores each customer on three axes and combines them into a segment:

| Axis | Definition | Direction |
|---|---|---|
| **Recency** | Days since last purchase | Lower is better |
| **Frequency** | Distinct invoices | Higher is better |
| **Monetary** | Total revenue | Higher is better |

Registered customers only. Recency is measured from the day after the last date in the dataset, so the most recent buyer scores 1 rather than 0.

### 3.1 Build the RFM table

````python
rfm_raw = con.execute("""
    SELECT
        CustomerID,
        DATE_DIFF('day',
            MAX(CAST(InvoiceDate AS DATE)),
            (SELECT MAX(CAST(InvoiceDate AS DATE)) + INTERVAL '1 day' FROM registered)
        ) AS Recency,
        COUNT(DISTINCT Invoice) AS Frequency,
        SUM(Revenue) AS Monetary
    FROM 
        registered
    WHERE 
        is_cancellation = 0 AND Quantity > 0
    GROUP BY 
        CustomerID
""").fetchdf()

rfm_raw
````

Distribution of each axis before scoring. The gap between median and mean signals how skewed each axis is.

````python
rfm_columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

query_parts = []

for column in rfm_columns:
    piece = f"""
    SELECT 
        '{column}' as column_name,
        COUNT(*) as total,
        MIN({column}) as min,
        QUANTILE_CONT({column},0.25) as qntile_25,
        QUANTILE_CONT({column},0.5) as median,
        QUANTILE_CONT({column},0.75) as qntile_75,
        MAX({column}) as max,
        ROUND(AVG({column}),2) as mean  
    FROM 
        rfm_raw
    """
    query_parts.append(piece)
rfm_distribution_query = " UNION ALL ".join(query_parts)
````

````python
rfm_distribution_df = con.execute(rfm_distribution_query).fetchdf()
rfm_distribution_df
````

### 3.2 Score and label

Each axis is split into quintiles with `NTILE(5)` so scores are relative to this customer base rather than to fixed thresholds. The seven segment labels are applied in priority order with `Needs Attention` catching anything that matches none of the rules.

````python
rfm_segments = con.execute("""
    WITH rfm_scored AS (
        SELECT *,
            NTILE(5) OVER (ORDER BY Recency  DESC) AS R_Score,
            NTILE(5) OVER (ORDER BY Frequency ASC) AS F_Score,
            NTILE(5) OVER (ORDER BY Monetary  ASC) AS M_Score
        FROM rfm_raw
    )
    SELECT *,
        CAST(R_Score AS VARCHAR) || 
        CAST(F_Score AS VARCHAR) || 
        CAST(M_Score AS VARCHAR) AS RFM_Score,
        CASE
            WHEN R_Score >= 4 AND F_Score >= 4 AND M_Score >= 4 THEN 'Champions'
            WHEN R_Score >= 3 AND F_Score >= 3 AND M_Score >= 3 THEN 'Loyal'
            WHEN R_Score >= 4 AND F_Score <= 2 THEN 'New Customers'
            WHEN R_Score >= 3 AND F_Score <= 2 THEN 'Promising'
            WHEN R_Score <= 2 AND F_Score >= 3 AND M_Score >= 3 THEN 'At Risk'
            WHEN R_Score <= 2 AND F_Score <= 2 THEN 'Lost'
            ELSE 'Needs Attention'
        END AS Segment
    FROM 
        rfm_scored
""").fetchdf()
rfm_segments
````

### 3.3 Segment sizes and value

**Question:** How do customer counts and revenue split across segments?

````python
con.execute("""
        SELECT
            Segment,
            COUNT(*) AS Customers,
            SUM(Frequency) AS Orders,
            ROUND(SUM(Frequency) * 100.0 / SUM(SUM(Frequency)) OVER (), 1) AS Order_Share_Pct,
            ROUND(SUM(Monetary), 2) AS Total_Revenue,
            ROUND(AVG(Monetary), 2) AS Avg_Revenue
        FROM 
            rfm_segments
        GROUP BY 
            Segment
        ORDER BY 
            Total_Revenue DESC
""").fetchdf()
````

````python
rfm_summary = con.execute("""
    SELECT
        Segment,
        COUNT(*) AS Customers,
        ROUND(SUM(Monetary), 2) AS Total_Revenue,
        ROUND(AVG(Monetary), 2) AS Avg_Revenue
    FROM 
        rfm_segments
    GROUP BY 
        Segment
    ORDER BY 
        Total_Revenue DESC
""").fetchdf()


top_seg = rfm_summary['Segment'][0]
top_cust_pct = round(rfm_summary['Customers'][0] / rfm_summary['Customers'].sum() * 100)
top_rev_pct = round(rfm_summary['Total_Revenue'][0] / rfm_summary['Total_Revenue'].sum() * 100)

rfm_summary
````

### 3.4 How concentrated is the top segment?

**Question:** Is Champions revenue spread across the segment or held by a handful of accounts?

````python
champion_concentration = con.execute("""
    WITH champions AS (
        SELECT
            CustomerID,
            Monetary,
            ROW_NUMBER() OVER (ORDER BY Monetary DESC) AS revenue_rank,
            SUM(Monetary) OVER () AS segment_revenue
        FROM rfm_segments
        WHERE Segment = 'Champions'
    )
    SELECT
        COUNT(*) AS champions,
        ROUND(MAX(segment_revenue), 0) AS champions_revenue,
        ROUND(AVG(Monetary), 2) AS mean_champion,
        ROUND(MEDIAN(Monetary), 0) AS median_champion,
        ROUND(SUM(CASE WHEN revenue_rank <= 5  THEN Monetary ELSE 0 END)
              / MAX(segment_revenue) * 100, 1) AS top_5_share_pct,
        ROUND(SUM(CASE WHEN revenue_rank <= 25 THEN Monetary ELSE 0 END)
              / MAX(segment_revenue) * 100, 1) AS top_25_share_pct
    FROM champions
""").fetchdf()

champion_concentration
````

> **Insight — concentration within Champions**
>
> * The segment is heavily skewed: half of all Champions generate less than £3,814, while the mean is more than double that.
> * Just 25 customers hold roughly a third of the segment's revenue.

**Question:** Does the whole customer base have a similar concentration?

````python
con.execute("""

    WITH customer_year AS (
        SELECT
            CASE
                WHEN CAST(InvoiceDate AS DATE) >= DATE '2009-12-01'
                 AND CAST(InvoiceDate AS DATE) <  DATE '2010-12-01' THEN 'Year 1: Dec09-Nov10'
                WHEN CAST(InvoiceDate AS DATE) >= DATE '2010-12-01'
                 AND CAST(InvoiceDate AS DATE) <  DATE '2011-12-01' THEN 'Year 2: Dec10-Nov11'
            END AS trading_year,
            CustomerID,
            SUM(Revenue) AS customer_revenue
        FROM 
            registered
        WHERE is_cancellation = 0 AND Quantity > 0
        GROUP BY 
            trading_year, 
            CustomerID
    ),
    ranked AS (
        SELECT
            *,
            ROW_NUMBER() OVER (PARTITION BY trading_year
                               ORDER BY customer_revenue DESC) AS revenue_rank,
            SUM(customer_revenue) OVER (PARTITION BY trading_year) AS year_revenue
        FROM 
            customer_year
        WHERE 
            trading_year IS NOT NULL
    )
    SELECT
        trading_year,
        COUNT(*) AS active_customers,
        ROUND(SUM(CASE WHEN revenue_rank <= 10 THEN customer_revenue ELSE 0 END)
              / MAX(year_revenue) * 100, 1) AS top_10_share_pct,
        ROUND(SUM(CASE WHEN revenue_rank <= 50 THEN customer_revenue ELSE 0 END)
              / MAX(year_revenue) * 100, 1) AS top_50_share_pct
    FROM ranked
    GROUP BY 
        trading_year
    ORDER BY 
        trading_year
    
""").fetchdf()
````

> **Insight — concentration overall**
>
> Concentration exists but is stable between the two years.

## 4. Customer acquisition

### 4.1 New versus returning

**Question:** How many customers each month are new and how much is repeat business?

````python
con.execute("""
    WITH sales AS (
        SELECT
            CustomerID,
            DATE_TRUNC('month', CAST(InvoiceDate AS DATE)) AS order_month
        FROM 
            registered
        WHERE is_cancellation = 0 AND Quantity > 0
    ),
    first_order AS (
        SELECT
            CustomerID,
            MIN(order_month) AS cohort_month
        FROM 
            sales
        GROUP BY 
            CustomerID
    ),
    monthly AS (
        SELECT
            s.order_month,
            COUNT(DISTINCT s.CustomerID) AS active_customers,
            COUNT(DISTINCT CASE WHEN f.cohort_month = s.order_month
                                THEN s.CustomerID END) AS new_customers
        FROM 
            sales s
        JOIN first_order f ON s.CustomerID = f.CustomerID
        GROUP BY 
            s.order_month
    )
    SELECT
        order_month,
        active_customers,
        new_customers,
        active_customers - new_customers AS returning_customers,
        ROUND(new_customers * 100.0 / active_customers, 1) AS pct_new
    FROM 
        monthly
    WHERE order_month >  DATE '2009-12-01' AND order_month <  DATE '2011-12-01'
    ORDER BY 
        order_month
""").fetchdf()
````

### 4.2 Does first-order value predict whether they come back?

Customers are split into four equal groups by first-order value and their return rates compared. Anyone whose first purchase falls in the last three months of the dataset is excluded as they have not had a fair chance to return yet and including them would understate the return rate for recent cohorts.

````python
con.execute("""
    WITH sales AS (
        SELECT
            CustomerID,
            Invoice,
            Revenue,
            InvoiceDate,
            FIRST_VALUE(Invoice) OVER (PARTITION BY CustomerID ORDER BY InvoiceDate, Invoice) AS first_invoice
        FROM 
            registered
        WHERE 
            is_cancellation = 0 AND Quantity > 0
    ),
    per_customer AS (
        SELECT
            CustomerID,
            COUNT(DISTINCT Invoice) AS orders,
            MIN(CAST(InvoiceDate AS DATE)) AS first_purchase_date,
            SUM(CASE WHEN Invoice = first_invoice THEN Revenue ELSE 0 END) AS first_order_value
        FROM 
            sales
        GROUP BY 
            CustomerID
    ),
    eligible AS (
        SELECT *
        FROM 
            per_customer
        WHERE 
            first_purchase_date < (SELECT MAX(CAST(InvoiceDate AS DATE)) FROM registered) - INTERVAL 3 MONTH
    ),
    bucketed AS (
        SELECT
            *,
            NTILE(4) OVER (ORDER BY first_order_value) AS quartile
        FROM eligible
    )
    SELECT
        quartile,
        ROUND(MIN(first_order_value), 0) AS min_first_order,
        ROUND(MAX(first_order_value), 0) AS max_first_order,
        COUNT(*) AS customers,
        ROUND(100.0 * SUM(CASE WHEN orders > 1 THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_returned
    FROM 
        bucketed
    GROUP BY 
        quartile
    ORDER BY 
        quartile
""").fetchdf()
````

> **Insight — acquisition quality**
>
> * The return rate rises at every step, from 65.5% in the lowest first-order quartile to 84.9% in the highest.
> * The sharpest jump is from the bottom quartile to the second (65.5% → 75.7%). Very low first-order spenders are the group least likely to come back. Above roughly £167 the return rates are high and similar.

### 6.2 Most returned products

**Question:** Which products have the highest returns and is that volume driven or a quality signal? Units returned are shown next to units sold and orders placed so the two can be judged together.

````python
# Top 10 Cancelled Products By Quantity Returned
con.execute("""
    WITH product_stats AS (
        SELECT
            StockCode,
            Description,
            SUM(CASE WHEN is_cancellation = 1 THEN ABS(Quantity) ELSE 0 END) AS units_returned,
            SUM(CASE WHEN is_cancellation = 0 AND Quantity > 0 THEN Quantity ELSE 0 END) AS units_sold,
            COUNT(DISTINCT CASE WHEN is_cancellation = 0 AND Quantity > 0 THEN Invoice END) AS orders_placed
        FROM 
            cleaned
        GROUP BY 
            StockCode,
            Description
    )
    SELECT
        Description,
        units_returned,
        units_sold,
        orders_placed
    FROM 
        product_stats
    ORDER BY 
        units_returned DESC
    LIMIT 10
""").fetchdf()
````

## 7. Time patterns

**Question:** Which days and hours generate the most revenue?

### 7.1 Revenue by day of week

````python
revenue_day = con.execute("""
    WITH day_of_week AS (
        SELECT
            *,
            CASE
                WHEN DayOfWeek = 0 Then 'Sunday'
                WHEN DayOfWeek = 1 Then 'Monday'
                WHEN DayOfWeek = 2 Then 'Tuesday'
                WHEN DayOfWeek = 3 Then 'Wednesday'
                WHEN DayOfWeek = 4 Then 'Thursday'
                WHEN DayOfWeek = 5 Then 'Friday'
                ELSE 'Saturday'
            END AS Days
        FROM
            cleaned
        WHERE 
            is_cancellation = 0 AND Quantity > 0
    )
    SELECT
        Days,
        Sum(Revenue) as Revenue
    FROM 
        day_of_week
    GROUP BY
        Days
    ORDER BY
        Revenue DESC
""").fetchdf()

revenue_day
````

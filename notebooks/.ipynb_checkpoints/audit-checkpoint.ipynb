## 1. Setup

```python
import duckdb
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', 50)

# Load both sheets and combine into 1 dataset
df1 = pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2009-2010')
df2 = pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2010-2011')
raw = pd.concat([df1, df2], ignore_index=True)


con = duckdb.connect()
con.register('raw_retail', raw)
print(f"Loaded {len(raw):,} rows into DuckDB")
```

## 3. Missing & Duplicate Values

```python
# Number of null values in the invoice column
con.execute("""
    SELECT 
        'invoice' as column_name,
        SUM(CASE WHEN invoice IS NULL THEN 1 ELSE 0 END) AS "Number of Nulls",
        ROUND(SUM(CASE WHEN "invoice" IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS "Pct_Nulls"
    FROM 
        raw_retail    
""").fetchdf()
```

```python
# Automating sql query above to handle multiple columns

df = con.execute("DESCRIBE raw_retail").fetchdf()

column_names = df['column_name'].tolist()

query_parts = []

for column_name in column_names:
    piece = f"""
    SELECT 
        '{column_name}' as column_name,
        SUM(CASE WHEN "{column_name}" IS NULL THEN 1 ELSE 0 END) AS "Number of Nulls",
        ROUND(SUM(CASE WHEN "{column_name}"  IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS "Pct_Nulls"
    FROM 
        raw_retail    
    """
    query_parts.append(piece)
missing_values_query = " UNION ALL ".join(query_parts)
```

```python
missing_values_df = con.execute(missing_values_query).fetchdf()
missing_values_df
```

```python
customer_count = con.execute("""
    SELECT 
        count(DISTINCT "customer ID") as customer_count
    FROM 
        raw_retail
""").fetchdf()['customer_count'][0]
```

```python
missing_customer_count = con.execute("""
    SELECT 
        count(*) as count
    FROM 
        raw_retail  
    WHERE 
        "Customer ID" IS NULL
""").fetchdf()['count'][0]
```

#### INSIGHT: 
#### • Only the description and customer ID columns have missing values. 0.41% of description have nulls but nearly a quarter of customer ID values are null. This could be transactions from those not registered as customers such as guests.

#### • Rows with null customer ID contribute to revenue but cannot be used in customer segmentation analysis

### Duplicate Rows

```python
# The rows which have duplicates and how many times repeated
con.execute("""
    SELECT
        Invoice, StockCode, Description, Quantity,
        InvoiceDate, Price, "Customer ID", Country,
        COUNT(*) AS times_repeated
    FROM 
        raw_retail
    GROUP BY
        Invoice, StockCode, Description, Quantity,
        InvoiceDate, Price, "Customer ID", Country
    HAVING COUNT(*) > 1
    ORDER BY 
        times_repeated DESC
""").fetchdf()
```

```python
# How much revenue do the duplicated rows generate
row_duplicate_revenue = con.execute("""
    WITH numbered AS (
        SELECT *,
            ROW_NUMBER() OVER (
                PARTITION BY Invoice, StockCode, Description, Quantity,
                             InvoiceDate, Price, "Customer ID", Country
            ) AS rn
        FROM 
            raw_retail
    )
    SELECT
        COUNT(*) AS excess_duplicate_rows,
        ROUND(SUM(Quantity * Price), 2) AS revenue_at_stake
    FROM 
        numbered
    WHERE 
        rn > 1
""").fetchdf()
row_duplicate_revenue
```

#### INSIGHT: 
#### • 34335 rows are exact duplicates across every column. These are 2% of the total and carry a revenue of £431717
#### • No line item ID exists so it's difficult to tell if these are extraction errors or genuine repeated entries
#### • Since they make up a small share of both rows and revenue, they will be removed rather than risk counting the same sale twice

## 4.Distributions

```python
# Stock Code Distribution
con.execute("""
   SELECT
        LENGTH(CAST(StockCode AS VARCHAR)) AS code_length,
        COUNT(*) AS rows,
        COUNT(DISTINCT StockCode) AS distinct_codes,
        ANY_VALUE(StockCode) AS example
    FROM 
        raw_retail
    GROUP BY 
        code_length
    ORDER BY 
        rows DESC
""").fetchdf()
```

```python
# Stock codes are mostly made up of numbers or numbers + upper case letters
# Counting how many stock codes have lower case letters
con.execute("""
    SELECT 
        COUNT(DISTINCT StockCode) AS raw_distinct
    FROM 
        raw_retail
    WHERE 
        StockCode != UPPER(CAST(StockCode AS VARCHAR))
""").fetchdf()
```

```python
# How many stock codes have lower case variants
con.execute("""
    SELECT
        UPPER(CAST(StockCode AS VARCHAR)) AS stock_code_upper,
        COUNT(DISTINCT StockCode) AS variants,
        COUNT(*) AS total_rows
    FROM 
        raw_retail
    GROUP BY 
        stock_code_upper
    HAVING 
        COUNT(DISTINCT StockCode) > 1
    ORDER BY 
        total_rows DESC
""").fetchdf()
```

```python
# Checking whether every stock code maps to exactly one description

con.execute("""
    SELECT
        StockCode,
        COUNT(DISTINCT Description) AS description_variants
    FROM 
        raw_retail
    WHERE
        Description IS NOT NULL
    GROUP BY 
        StockCode
    HAVING 
        COUNT(DISTINCT Description) > 1
    ORDER BY 
        description_variants DESC
""").fetchdf()
```

```python
# StockCode 20713 as an example
con.execute("""
    SELECT
        StockCode,
        Description
    FROM 
        raw_retail
    WHERE
        StockCode = '20713' AND Description IS NOT NULL
""").fetchdf()
```

```python
# Counting number of jumbo bag owls which is the description that seems to represent the stock code 20713
con.execute("""
    SELECT
        Description,
        count(*) as count
    FROM 
        raw_retail
    GROUP BY
        Description
    HAVING
        Description = 'JUMBO BAG OWLS'
""").fetchdf()
```

```python
# Checking the 8 rows that don't seem to have JUMBO BAG OWLS description
con.execute("""
    SELECT
        Description,
        COUNT(*) AS rows
    FROM 
        raw_retail
    WHERE 
        StockCode = '20713' AND Description IS NOT NULL
    GROUP BY 
        Description
    ORDER BY 
        rows DESC
""").fetchdf()
```

#### INSIGHT: 
#### • Stock codes are typically 5 or 6 characters long, mostly composed of numbers or numbers + letters.
#### • Majority of stock codes with letters have uppercased letters. 183 stock codes have lower case letters and 173 of these have upper case version                   representing the same product. Should convert all the letters in stock codes to uppercase
#### • Some stock codes are paired with descriptions that don't have the product name

```python
# Price & Quantity Distribution
numerical_columns = ['Quantity', 'Price']

numeric_query_parts = []

for column in numerical_columns:
    piece = f"""
    SELECT 
        '{column}' as column_name,
        COUNT(*) as total,
        ROUND(AVG({column}),2) as mean,
        MIN({column}) as min,
        QUANTILE_CONT({column},0.25) as qntile_25,
        QUANTILE_CONT({column},0.5) as median,
        QUANTILE_CONT({column},0.75) as qntile_75,
        MAX({column}) as max,
    FROM 
        raw_retail    
    """
    numeric_query_parts.append(piece)
numeric_columns_query = " UNION ALL ".join(numeric_query_parts)
```

```python
numeric_columns_df = con.execute(numeric_columns_query).fetchdf()
numeric_columns_df
```

## 5.Date/Time Analysis

```python
# Full date range and distinct days

date_range = con.execute("""
    SELECT
    MIN(InvoiceDate) AS earliest_date,
    MAX(InvoiceDate) AS latest_date,
    COUNT(DISTINCT CAST(InvoiceDate AS DATE)) AS unique_days
    FROM raw_retail;
""").fetchdf()

date_range
```

```python
earliest_date = date_range['earliest_date'][0]
latest_date = date_range['latest_date'][0]
unique_days = date_range['unique_days'][0]
```

```python
con.execute("""
    SELECT
    DATE_TRUNC('month', InvoiceDate) AS month,
    COUNT(*) AS no_observations,
    COUNT(DISTINCT CAST(InvoiceDate AS DATE)) AS trading_days
    FROM raw_retail
    GROUP BY month
    ORDER BY month ASC;
""").fetchdf()
```

#### INSIGHT: The dataset covers 604 unique days from 1st december 2009 to 9th december 2011. No months are missing. The months have similar number of trading days in the 20s except for december 2011 that only had 8 trading days but the number of observations from that month was still in line with rest of the months
